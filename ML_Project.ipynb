# ***Feature Engineering***

```python
top_countries = df['Country'].value_counts().index[:10].tolist()
top_countries
```

```python
def group_countries(country):
  if country in top_countries:
    return country
  else:
    return 'Other'
```

```python
df['Grouped_countries']= df['Country'].apply(group_countries)
df['Grouped_countries'].value_counts()
```

# ***Train/Test Split***

```python
skew_feature = ['Study_Hours']
numeric_feature = ['Age','Avg_Daily_Usage_Hours', 'Sleep_Hours_Per_Night',
                   'Physical_Activity_Hours', 'Daily_Unlocks']
ordinal_feature = ['Stress_Level']
nominal_feature = ['Gender', 'Academic_Level','Grouped_countries', 'Most_Used_Platform',
                   'Purpose_Of_Use']

Feature_col = skew_feature + numeric_feature + ordinal_feature + nominal_feature
Target_col = ['Mental_Health_Score']

X = df[Feature_col]
y = df[Target_col]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

# ***Building Pipeline***

```python
# Skewed Feature
skew_pipeline = Pipeline(steps=[
    ('log_transform', FunctionTransformer(np.log1p)),
    ('scaler', StandardScaler())
])

# Numeric Feature
num_pipeline = Pipeline(steps=[
    ('scaler', StandardScaler())
])

# Ordinal Feature
ordinal_pipeline = Pipeline(steps=[
    ('ordinal_encoder', OrdinalEncoder(categories=[['Low', 'Medium', 'High', 'Very High']]))
    ])

# Nominal Feature
nominal_pipeline = Pipeline(steps=[
    ('onehot_encoder', OneHotEncoder(handle_unknown='ignore'))
])

#Column Transormers
preprocessor = ColumnTransformer(transformers=[
    ('skew_pipeline', skew_pipeline, skew_feature),
    ('num_pipeline', num_pipeline, numeric_feature),
    ('ordinal_pipeline', ordinal_pipeline, ordinal_feature),
    ('nominal_pipeline', nominal_pipeline, nominal_feature)
])
```

# ***Building the models***

```python
# 1. Flatten targets to 1D arrays to prevent warnings
y_train_flat = y_train.values.ravel() if hasattr(y_train, "values") else np.ravel(y_train)
y_test_flat = y_test.values.ravel() if hasattr(y_test, "values") else np.ravel(y_test)

# 2. Define multiple scoring metrics for cross-validation
# Note: Scikit-learn uses negative values for error metrics so higher is always better
scoring = {
    'r2': 'r2',
    'mae': 'neg_mean_absolute_error',
    'mse': 'neg_mean_squared_error'
}

# 3. Model configurations
models = {
    'LinearRegression': (LinearRegression(), {}),

    'SVR': (SVR(), {
        'model__C': [0.1, 1, 10],
        'model__gamma': [1, 0.1, 0.01],
        'model__epsilon': [0.1, 0.01, 0.001]
    }),

    'DecisionTreeRegressor': (DecisionTreeRegressor(random_state=42), {
        'model__max_depth': [None, 5, 10],
        'model__splitter': ['best', 'random']
    }),

    'RandomForestRegressor': (RandomForestRegressor(random_state=42), {
        'model__n_estimators': [10, 100, 200],
        'model__max_depth': [None, 5, 10]
    }),

    'KNeighborsRegressor': (KNeighborsRegressor(), {
        'model__n_neighbors': np.arange(3, 50, 2),
        'model__weights': ['uniform', 'distance']
    })
}

results_list = []
trained_pipelines = {}

# 4. Run GridSearchCV across all models
for name, (model, params) in models.items():
    print(f"Running GridSearchCV on {name}...")
    
    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('model', model)
    ])

    grid_search = GridSearchCV(
        estimator=pipeline,
        param_grid=params,
        cv=5,
        scoring=scoring,
        refit='r2',     # Refits the best estimator using R2 score
        n_jobs=4        # Fixed worker count to avoid BrokenProcessPool memory issues
    )

    grid_search.fit(X_train, y_train_flat)

    # Save the best fitted estimator
    best_pipe = grid_search.best_estimator_
    trained_pipelines[name] = best_pipe

    # Evaluate the best pipeline on UNSEEN TEST DATA
    y_test_pred = best_pipe.predict(X_test)
    test_r2 = r2_score(y_test_flat, y_test_pred)
    test_mae = mean_absolute_error(y_test_flat, y_test_pred)
    test_mse = mean_squared_error(y_test_flat, y_test_pred)

    # Extract CV validation scores
    cv_idx = grid_search.best_index_
    cv_r2 = grid_search.cv_results_['mean_test_r2'][cv_idx]
    cv_mae = -grid_search.cv_results_['mean_test_mae'][cv_idx]  # Invert negative sign
    cv_mse = -grid_search.cv_results_['mean_test_mse'][cv_idx]

    results_list.append({
        'Model': name,
        'CV_R2': cv_r2,
        'CV_MAE': cv_mae,
        'CV_MSE': cv_mse,
        'Test_R2': test_r2,
        'Test_MAE': test_mae,
        'Test_MSE': test_mse,
        'Best_Params': grid_search.best_params_
    })

# 5. Display comparative DataFrame sorted by Test R2
results_df = pd.DataFrame(results_list).sort_values(by='Test_R2', ascending=False).reset_index(drop=True)
print("\n--- MODEL COMPARISON TABLE ---")
print(results_df[['Model', 'Test_R2', 'Test_MAE', 'Test_MSE', 'CV_R2']])

# 6. Retrieve the overall winner
winner_name = results_df.loc[0, 'Model']
winner_pipeline = trained_pipelines[winner_name]

print("\n" + "="*45)
print(f"🏆 OVERALL WINNER: {winner_name}")
print(f"Test R2:  {results_df.loc[0, 'Test_R2']:.4f} (Higher is better)")
print(f"Test MAE: {results_df.loc[0, 'Test_MAE']:.4f} (Lower is better)")
print(f"Test MSE: {results_df.loc[0, 'Test_MSE']:.4f} (Lower is better)")
print(f"Best Parameters: {results_df.loc[0, 'Best_Params']}")
print("="*45)
```

```python
joblib.dump(winner_pipeline, 'Mental_health_model.pkl')
```
